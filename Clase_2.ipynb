ARITMETICA Y OPERADORES CON ARREGLOS

Los arreglos de numpy son compatibles con los diversos operadores aritmeticos y logicos de python

```python
matriz_1 = np.array([[1,2,3],
                    [4,0,-5],
                    [-6,-7,-8]])
matriz_1
```

Todas las operaciones siguientes seran aplicadas a todos y cada uno de los elementos del arreglo

```python
matriz_1 + 100
```

```python
matriz_1[2,:]*3
```

```python
3/matriz_1
```

```python
matriz_1/2
```

```python
matriz_1 < 1
```

```python
matriz_1.shape
```

```python
matriz_2 = np.array([1,2,3])
```

```python
matriz_2.shape
```

```python
matriz_1 * matriz_2
```

```python
matriz_3 = np.array([[2], [4], [6]])
```

```python
matriz_3
```

```python
matriz_3.shape
```

```python
matriz_3 * matriz_1
```
